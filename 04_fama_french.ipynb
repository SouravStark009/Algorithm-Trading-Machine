Now lets attach the weights to our features using Rolling OLS

```python
factor_data
```

```python
from statsmodels.regression.rolling import RollingOLS
import statsmodels.api as sm
```

```python
factor_data.groupby(level=1,group_keys=False).apply(lambda x: RollingOLS(endog=x['return_1m'],
                                                                         exog=sm.add_constant(x.drop('return_1m',axis=1)),
                                                                        window=min(24,x.shape[0]),
                                                                        min_nobs=len(x.columns)+1)
   .fit(params_only=True) 
   .params.drop('const',axis=1))
```

```python
betas = (factor_data.groupby(level=1,group_keys=False).apply(lambda x: RollingOLS(endog=x['return_1m'],
                                                                         exog=sm.add_constant(x.drop('return_1m',axis=1)),
                                                                        window=min(24,x.shape[0]),
                                                                        min_nobs=len(x.columns)+1)
.fit(params_only=True)
.params
.drop('const',axis=1)))

betas
```

```python
betas.groupby('ticker').shift()
```

```python
betas.index
```

```python
factors = ['Mkt-RF', 'SMB', 'HML', 'RMW', 'CMA']

data = (data.join(betas.groupby('ticker').shift()))
```

Yeh .shift() isliye karna pada kyuki manlo august month ka jo data woh data toh kam sepetember month ke analysis ke liye kam aayega na toh actually august ka kam september ke decisions ko shape karta toh august ka data september ke liye meaningful hai na ki august ke liye isliye ye chiz model ko smjhane ke liye hume data ko ek position aage shift karna pada tha warna agr nhi karenge toh bhut complex ho jayega ml model ke liye smjhna  

```python
data.loc[:, factors] = data.groupby('ticker', group_keys=False)[factors].apply(lambda x: x.fillna(x.mean()))
```
